```python
import numpy as np
CATEGORIES = ['Cat', 'Dog']


def image(path):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
    new_arr = cv2.resize(img, (60, 60))
    new_arr = np.array(new_arr)
    new_arr = new_arr.reshape(-1, 60, 60, 1)
    return new_arr

import shutil
shutil.copy('3x3x64-catvsdog.model', 'model.h5')
model = keras.models.load_model('model.h5')
```

```python
import os
folder = r'C:\Users\hp\Desktop\cat n dog\dogscats\dogscats\train\cats'
print(os.listdir(folder)[:5])
```

```python
img_path = os.path.join(folder, 'cat.1.jpg')
print(os.path.exists(img_path))
prediction = model.predict([image(img_path)])
print(CATEGORIES[prediction.argmax()])
```
